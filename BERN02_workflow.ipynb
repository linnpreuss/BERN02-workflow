# Analysis o the Palmer Penguins Dataset

This notebook presents a producible analysis of the Palmer Penguins dataset. The aim is to investigate the relationship between penguin 
flipper length and the body mass.

# Requirements
This workflow uses Python and the following packages:
- pandas
- seaborn
- matplotlib
- numpy

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd

df= sns.load_dataset("penguins")
df=df.dropna(subset=["flipper_length_mm", "body_mass_g", "species"])
```

# Mean body mass
The arithmetic mean is 

$$ \bar{x} = \frac{1}{N}\sum_{i=1}^{N}x_i. $$

```python
#Calculate mean
df.groupby("species")["body_mass_g"].mean()
```

```python
#Scatter plot
#Each point represents one penguin. Colour indicates the species. 

sns.scatterplot(
    data=df,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species"
)

plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.title("Body mass by penguin species")
plt.show()
```
